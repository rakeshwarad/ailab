```python
def occurs_check(var, expr, subst):
    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(occurs_check(var, item, subst) for item in expr[1:])

    if isinstance(expr, str) and expr in subst:
        return occurs_check(var, subst[expr], subst)

    return False


def unify(x, y, subst=None):
    if subst is None:
        subst = {}

    if x == y:
        return subst

    if isinstance(x, str) and x.islower():
        if x in subst:
            return unify(subst[x], y, subst)

        if occurs_check(x, y, subst):
            return None

        subst[x] = y
        return subst

    if isinstance(y, str) and y.islower():
        return unify(y, x, subst)

    if isinstance(x, tuple) and isinstance(y, tuple):
        if x[0] != y[0] or len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            result = unify(a, b, subst)

            if result is None:
                return None

            subst = result

        return subst

    return None


def test_case(name, x, y):
    print("\n" + name)
    result = unify(x, y)

    if result is None:
        print("Unification Failed")
    elif result:
        print("Unification Successful")
        print("Substitution:", result)
    else:
        print("No substitution required (NIL)")


test_case("Case 1: Identical expressions",
          ("P", "x"), ("P", "x"))

test_case("Case 2: L1 is variable",
          "x", "Riya")

test_case("Case 3: L2 is variable",
          "Riya", "y")

test_case("Case 4: Occurs check",
          "x", ("F", "x"))

test_case("Case 5: Different predicates",
          ("Eats", "x"), ("Likes", "x"))

test_case("Case 6: Different arguments",
          ("P", "x", "y"), ("P", "x"))

test_case("Case 7: Multiple arguments",
          ("P", "x", "Apple", "z"),
          ("P", "Riya", "y", "Mango"))

test_case("Case 8: Conflicting substitutions",
          ("P", "x", "x"), ("P", "John", "Riya"))
```
